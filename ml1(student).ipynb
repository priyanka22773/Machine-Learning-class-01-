```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
```

```python
df.drop(columns='Performance Index')
```

```python
x = df.drop(columns='Performance Index') #i/p()
y = df['Performance Index']
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(x,y,train_size=0.8,random_state=42)
```

**random_state is a parameter used in Machine Learning to control randomness.
It ensures that you get the same output every time you run your code.**

```python
#model creation
model = LinearRegression()
```

```python
df
```

```python
#model training
model.fit(xtrain,ytrain) # model learns the pattern from xtrain -->i/p,ytrain-->o/p

#pattern --> mx+c(stright-line)
```

**o/p will be stored in object(model)**
